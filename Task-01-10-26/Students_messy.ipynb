#### Removing dublicates

```python
df = df.drop_duplicates()
```

#### Removing extra spaces

```python
df["Name"] = df["Name"].str.strip()
df["Gender"] = df["Gender"].str.strip()
df["City"] = df["City"].str.strip()
df["Email"] = df["Email"].str.strip()
```

```python
df['Gender']= df['Gender'].str.title()
df['City'] = df['City'].str.title()
```

```python
age = df["Age"].median()
print(age)
```

#### Checking and Handling Invalid values

```python
df[["Age", "Maths", "Physics", "Attendance"]]
```

##### Set valid ranges
Age should be between 15 and 100 

Maths should be between 0 and 100

Physics should be between 0 and 100

Attendance should be between 0 and 100

##### Find Invalid Age

```python
df.loc[~df['Age'].between(15,100), 'Age']
```

```python
df.loc[~df['Age'].between(15,100), 'Age'] = pd.NA
```

```python
df["Age"].isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"].isnull().sum()
df['Age']
```

##### Find invalid Attendance

```python
df.loc[~df["Attendance"].between(0, 100), "Attendance"] = pd.NA
```

```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
```

```python
df['Attendance'].isnull().sum()
```
